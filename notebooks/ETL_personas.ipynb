**ACERCA DE LA POBREZA EXTREMA**

Nuestro punto de partida fue este fragmento de una artiçulo del World Economic Forum con conclusiones de 2024 a 2024 que deciá los siguiente:

*6,9%, es decir, 800 millones de personas, siguen viviendo en condiciones de pobreza extrema. Esto supone solo 1,5 puntos porcentuales menos que en 2015 (8,4%) y se prevé que la tasa se mantenga prácticamente estancada durante los próximos cinco años.*

*La ONU afirma que es poco probable que se erradique la pobreza extrema como consecuencia de la lenta recuperación de la COVID-19, la volatilidad económica, las crisis climáticas y el lento crecimiento en regiones como el África subsahariana, donde vive la mayoría de los pobres del mundo.*

La pobreza son muchas cosas, y por lo que investigué su forma de medición no deja contentos a todo el mundo. La forma más usual de hacerlo es mediante el ingreso. Hay pobreza monetaria, y la multidimensional. Prácticamente, se determina por lo que puede pagar una persona mensualmente respecto de dos líneas: la de pobreza lp, que aparece en la descripción´de los campos, y la pobreza extrema, li (línea de indigencia).

El DANE determina el ingreso en dinero de un hogar y lo divide entre el número de personas del mismo. A ese número de personas se le llama "unidad de gasto per cápita". Si un hogar tiene 5 personas, la unidad de gasto de ese hogar se compone de 5 personas. El ingreso mensual se divide así entre ellos. Si Santiago o Sebastián vivieran solos, su unidad de gasto sería de 1 cada uno. Como no les entra nada, serían clasificados como pobres extremos (indigentes). Sin embargo, como dependen de su familia, se espera que la cantidad de dinero que entra a esa familia sea repartido "equitatiamente" y, por tanto, los dos no están en indigencia. Si bien no es objetivo de lo que estamos haciendo, sí les añado que la "pobreza multidimensional" permite añadir más cosas que el ingreso, para determinar que tan "cómoda" con su cotidianidad vive una persona, no solo con lo que tiene en las arcas (dinero).

Al igual que en Hogares, estos datos son una muestra: no se estudió a toda la población, sino a un número limitado de personas. La columna `fex_c` indica cuántas personas reales representa esa fila.
````python
#este fragmento de código, la definición de una clase, permite varias acciones con los datos:
class da:


  def __init__(self, df):
    self.df = df


  def GraficarVacios(self):
    plt.figure()
    sns.heatmap(self.df.isnull(), cbar=False)
    plt.show()

  def summary(self):
    print(self.df.describe())

  #equivalente en porcentaje
  def missing_values(self):
    return self.df.isnull().sum()*100/len(self.df)

  #por valores absolutos
  def missing_values_abs(self):
    return self.df.isnull().sum()

  def columns_by_type(self):
    print(self.df.dtypes)
````

## 3. Modelo dimensional (esquema galaxia / constelación)

`fact_personas` se suma como segunda tabla de hechos del mismo esquema galaxia que ya tiene `fact_hogares`. Por eso reutiliza las dimensiones conformadas `dim_geografia`, `dim_tiempo` y `dim_hogar` que ya viven en Postgres — **no se reconstruyen desde `Personas.csv`**, se traen sus llaves (`id_geografia`, `id_tiempo`, `id_hogar`) uniendo por las mismas columnas naturales (`dpto`/`dominio`/`ciudad_sinam`/`clase` y `mes`, y `directorio`/`secuencia_p` para el hogar).

Siguiendo los significados oficiales del diccionario del DANE, la separación entre atributos descriptivos y medidas queda así:

**`dim_persona`** — identifica y describe a la persona, no mide nada:
- `directorio`, `secuencia_p`, `orden` → llaves (las dos primeras enlazan con `dim_hogar`)
- `p3271` → sexo
- `p6040` → edad
- `p6050` → parentesco con el jefe del hogar
- `p3042` → nivel educativo más alto alcanzado
- `condicion_laboral` → se arma combinando `oc`/`des`/`fft` (Ocupado / Desocupado / Fuera de la fuerza de trabajo), que en el CSV original son 3 columnas binarias mutuamente excluyentes
- `pet` → si está en edad de trabajar (booleano)

**`fact_personas`** — las medidas + llaves foráneas:
- `id_persona` (PK, y también FK hacia `dim_persona`, igual que `id_hogar` en `fact_hogares`), `id_hogar`, `id_geografia`, `id_tiempo`
- `ingtot` → ingreso total de la persona (la medida central)
- `p6800` → horas trabajadas a la semana
- `p6500` → ingreso del trabajo principal
- `p7070` → ingreso del segundo trabajo
- `p7045` → horas del segundo trabajo
- `fex_c`, `fex_dpto` → factores de expansión, para ponderar igual que en `fact_hogares`

> Nota: `oc` y `des` no estaban en el `encabezados` original de la sección de Extracción (solo se había traído `fft`), así que se añadieron ahí para poder construir `condicion_laboral`. ### Capa Oro

Última capa del medallón para personas. Aquí no se limpia nada nuevo: se derivan las columnas que el EDA necesita y que no conviene recalcular en cada consulta SQL.

**1. Etiquetas legibles** (`sexo`, `nivel_educativo`, `grupo_edad`) → viajan a `dim_persona`, igual que `categoria_tenencia`/`tipo_zona` viajan a las dimensiones del notebook de hogares. Así el EDA agrupa por texto directo en SQL, sin reconstruir el `CASE WHEN` en cada query. El diccionario de `p3042` es el oficial del DANE (archivo `niveles_educativos.csv`).

**2. `tiene_segundo_empleo`** (0/1 a partir de `p7040`) → viaja a `fact_personas`. Es la medida que responde la pregunta de multiempleo; como bandera numérica se puede ponderar con `SUM(tiene_segundo_empleo * fex_c) / SUM(fex_c)`, igual que `pobre`/`indigente` en `fact_hogares`.

**3. Columnas `_log`** (`log1p`) de las variables monetarias y de horas. El perfilado de outliers de la capa Plata mostró que `ingtot`, `p6500` y `p7070` son fuertemente asimétricas a la derecha: un promedio aritmético queda arrastrado por unos pocos ingresos extremos y deja de describir a la persona típica. Por eso se persiste la versión logarítmica **además** de la original, y los agregados del EDA se calculan como

```sql
EXP( SUM(col_log * fex_c) / SUM(fex_c) ) - 1
```

que es la **media geométrica ponderada**, robusta a los valores extremos. Se usa `log1p` (y no `log`) porque hay ceros legítimos: `log1p(0) = 0` en vez de `-inf`. Los nulos estructurales se mantienen nulos — no se convierten en 0, porque "no tiene segundo empleo" no es lo mismo que "gana 0 en su segundo empleo".

> Ninguna columna original se modifica ni se elimina: la capa Oro solo **agrega**. Todo lo que ya estaba (`ingtot`, `p6500`, `p6800`, `p7070`, `p7045`) sigue viajando tal cual a `fact_personas`, así que cualquier consulta escrita antes de este cambio sigue dando el mismo resultado.


````python
oro = plata.copy()

# --- 1. Etiquetas legibles (van a dim_persona) ---------------------------------
oro["sexo"] = oro["p3271"].map({1: "Hombre", 2: "Mujer"})

# Diccionario oficial DANE del campo p3042 (nivel educativo más alto alcanzado).
# Mismo contenido de niveles_educativos.csv; se deja embebido para que el notebook
# sea reproducible sin depender de un archivo auxiliar en Drive.
mapa_educacion = {
    1: "Ninguno",
    2: "Preescolar",
    3: "Básica primaria (1o - 5o)",
    4: "Básica secundaria (6o - 9o)",
    5: "Media académica (Bachillerato clásico)",
    6: "Media técnica (Bachillerato técnico)",
    7: "Normalista",
    8: "Técnica profesional",
    9: "Tecnológica",
    10: "Universitaria",
    11: "Especialización",
    12: "Maestría",
    13: "Doctorado",
    99: "No sabe, no informa",
}
oro["nivel_educativo"] = oro["p3042"].map(mapa_educacion)

# Grupos de edad: la edad exacta (p6040) se conserva en dim_persona; el bucket sirve
# para agrupar en SQL sin escribir rangos a mano en cada consulta.
oro["grupo_edad"] = pd.cut(
    oro["p6040"],
    bins=[-1, 5, 11, 17, 28, 45, 60, 200],
    labels=["0-5", "6-11", "12-17", "18-28", "29-45", "46-60", "61+"],
).astype(str)

# --- 2. Bandera de multiempleo (va a fact_personas) ---------------------------
# p7040: 1 = sí tenía otro trabajo o negocio, 2 = no. Los nulos son estructurales
# (personas que no están ocupadas), y se dejan nulos: las consultas de multiempleo
# filtran por condicion_laboral = 'Ocupado'.
oro["tiene_segundo_empleo"] = oro["p7040"].map({1: 1, 2: 0})

# --- 3. Columnas _log (van a fact_personas) -----------------------------------
# log1p mantiene los ceros legítimos y deja los nulos estructurales como nulos.
cols_log = {
    "ingtot": "ingtot_log",
    "p6500": "salario_principal_log",
    "p7070": "ingreso_segundo_empleo_log",
    "p6800": "horas_trabajadas_log",
    "p7045": "horas_segundo_empleo_log",
}
for origen, destino in cols_log.items():
    oro[destino] = np.log1p(oro[origen])

# Verificación: los _log no pueden inventar ni perder datos respecto de su origen
for origen, destino in cols_log.items():
    print(f"{origen:8s} -> {destino:28s} nulos: {oro[origen].isna().sum():>7,} / {oro[destino].isna().sum():>7,}")

print("\nMedia aritmética vs. media geométrica de ingtot (sin ponderar, solo ilustrativo):")
print(f"  aritmética: {oro['ingtot'].mean():,.0f}")
print(f"  geométrica: {np.expm1(oro['ingtot_log'].mean()):,.0f}")

oro[["sexo", "nivel_educativo", "grupo_edad", "tiene_segundo_empleo", "ingtot", "ingtot_log"]].head(10)
````

## 5. EDA — consultando la base de datos

A partir de acá no se vuelve a tocar ningún DataFrame de la extracción: todo sale de `fact_personas`, `dim_persona`, `fact_hogares` y `dim_geografia` en Postgres, con `pd.read_sql`. `dim_hogar` actúa de puente — es lo que permite llevar la condición de pobreza del **hogar** hasta cada **persona** sin haber duplicado las medidas del hogar en el modelo.

Dos reglas que se aplican en todas las consultas:

- **Ponderación obligatoria.** Cada fila representa a `fex_c` personas reales, así que las proporciones se calculan como `SUM(bandera * fex_c) / SUM(fex_c)` y nunca con `COUNT(*)`. En los cortes departamentales se usa `fex_dpto`, que es el factor calibrado para ese nivel.
- **Ingresos en media geométrica.** Las variables monetarias son asimétricas (capa Plata), así que se agregan como `EXP(SUM(col_log * fex_c) / SUM(fex_c)) - 1` sobre las columnas `_log` de la capa Oro.


### Pregunta 4 — ¿Las personas de hogares pobres están más en trabajos de baja remuneración o en múltiples empleos?

Se restringe a personas **ocupadas** (`condicion_laboral = 'Ocupado'`), que son las únicas para las que `p6500`/`p7040` tienen sentido. La condición del hogar se trae desde `fact_hogares` por `id_hogar`, y se parte en tres grupos excluyentes porque un hogar indigente también está marcado como pobre.

````python
query_p4 = """
WITH ocupados AS (
    SELECT
        CASE WHEN fh.indigente = 1 THEN 'Hogar en indigencia'
             WHEN fh.pobre = 1     THEN 'Hogar pobre (no indigente)'
             ELSE                       'Hogar no pobre' END AS condicion_hogar,
        COALESCE(fp.tiene_segundo_empleo, 0) AS segundo_empleo,
        fp.salario_principal_log,
        fp.fex_c
    FROM fact_personas fp
    JOIN dim_persona  dp ON dp.id_persona = fp.id_persona
    JOIN fact_hogares fh ON fh.id_hogar   = fp.id_hogar
    WHERE dp.condicion_laboral = 'Ocupado'
)
SELECT
    condicion_hogar,
    SUM(segundo_empleo * fex_c) / SUM(fex_c) * 100 AS pct_segundo_empleo,
    EXP(
        SUM(salario_principal_log * fex_c) FILTER (WHERE salario_principal_log IS NOT NULL)
        / SUM(fex_c)                       FILTER (WHERE salario_principal_log IS NOT NULL)
    ) - 1 AS salario_geometrico,
    SUM(fex_c) AS personas_representadas
FROM ocupados
GROUP BY condicion_hogar
ORDER BY salario_geometrico
"""
p4 = pd.read_sql(text(query_p4), engine)
print(p4)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

axes[0].bar(p4["condicion_hogar"], p4["salario_geometrico"], color="#16a085")
axes[0].set_ylabel("Salario del trabajo principal (media geométrica ponderada, $)")
axes[0].set_title("Remuneración del trabajo principal")
axes[0].tick_params(axis="x", rotation=20)

axes[1].bar(p4["condicion_hogar"], p4["pct_segundo_empleo"], color="#d35400")
axes[1].set_ylabel("% de ocupados con segundo empleo (ponderado)")
axes[1].set_title("Multiempleo")
axes[1].tick_params(axis="x", rotation=20)

plt.suptitle("Pregunta 4 — Remuneración y multiempleo según la condición de pobreza del hogar")
plt.tight_layout()
plt.show()
````

### Pregunta 5 — ¿Dónde hay más pobreza extrema, a nivel de personas?

La misma pregunta geográfica del notebook de hogares, pero con el grano de persona: un departamento puede tener pocos hogares indigentes y aun así muchas personas viviendo en ellos, porque los hogares indigentes tienden a ser más grandes. Aquí se pondera con `fex_dpto`, que es el factor de expansión calibrado para agregados departamentales.

````python
query_p5 = """
SELECT
    g.dpto,
    SUM(fh.indigente * fp.fex_dpto) / SUM(fp.fex_dpto) * 100 AS pct_personas_indigencia,
    SUM(fh.pobre     * fp.fex_dpto) / SUM(fp.fex_dpto) * 100 AS pct_personas_pobres,
    SUM(fp.fex_dpto) AS personas_representadas
FROM fact_personas  fp
JOIN fact_hogares   fh ON fh.id_hogar     = fp.id_hogar
JOIN dim_geografia  g  ON g.id_geografia  = fp.id_geografia
GROUP BY g.dpto
ORDER BY pct_personas_indigencia DESC
LIMIT 10
"""
p5 = pd.read_sql(text(query_p5), engine)
print(p5)

fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(p5["dpto"], p5["pct_personas_indigencia"], color="#c0392b")
ax.invert_yaxis()
ax.set_xlabel("% de personas en pobreza extrema (ponderado por fex_dpto)")
ax.set_title("Pregunta 5 — 10 departamentos con mayor pobreza extrema a nivel de personas")
plt.tight_layout()
plt.show()
````

### Pregunta 6 — ¿Quiénes están en pobreza extrema? (sexo y edad)

Tasa de indigencia ponderada cruzando `grupo_edad` × `sexo`, ambos atributos de `dim_persona`.

### Pregunta 8 — ¿Cómo se comportan los ingresos de las personas según la condición de pobreza de su hogar?

Dos miradas complementarias sobre la misma variable:

1. **Media geométrica ponderada** de `ingtot` por condición del hogar, calculada sobre `ingtot_log` — el resumen robusto que justifica toda la capa Oro.
2. **Distribución completa**, construida con `width_bucket` sobre `ingtot_log` y sumando `fex_c` en cada bin. El histograma se arma íntegramente en SQL (la base devuelve ~20 filas por grupo, no 770 mil), y el eje x se re-etiqueta con `expm1` para leerlo en pesos.

La media sola escondería lo importante: si dos grupos tienen medias distintas pero distribuciones superpuestas, el ingreso no separa tan bien como parece.

````python
query_p8_resumen = """
SELECT
    CASE WHEN fh.indigente = 1 THEN 'Hogar en indigencia'
         WHEN fh.pobre = 1     THEN 'Hogar pobre (no indigente)'
         ELSE                       'Hogar no pobre' END AS condicion_hogar,
    EXP(SUM(fp.ingtot_log * fp.fex_c) / SUM(fp.fex_c)) - 1 AS ingreso_geometrico,
    SUM(fp.ingtot * fp.fex_c) / SUM(fp.fex_c)              AS ingreso_aritmetico,
    SUM(fp.fex_c) AS personas_representadas
FROM fact_personas fp
JOIN fact_hogares  fh ON fh.id_hogar = fp.id_hogar
WHERE fp.ingtot > 0   -- se excluyen los ceros (menores, personas sin ingreso propio):
      -- la pregunta es sobre el nivel del ingreso de quien percibe alguno
GROUP BY condicion_hogar
ORDER BY ingreso_geometrico
"""
p8 = pd.read_sql(text(query_p8_resumen), engine)
print(p8)
print("\nLa distancia entre la media aritmética y la geométrica es la asimetría que motivó las columnas _log.")
````

````python
query_p8_hist = """
SELECT
    CASE WHEN fh.indigente = 1 THEN 'Hogar en indigencia'
         WHEN fh.pobre = 1     THEN 'Hogar pobre (no indigente)'
         ELSE                       'Hogar no pobre' END AS condicion_hogar,
    width_bucket(fp.ingtot_log, 8, 18, 20) AS bin,
    SUM(fp.fex_c) AS personas
FROM fact_personas fp
JOIN fact_hogares  fh ON fh.id_hogar = fp.id_hogar
WHERE fp.ingtot > 0
GROUP BY condicion_hogar, bin
ORDER BY condicion_hogar, bin
"""
hist = pd.read_sql(text(query_p8_hist), engine)

# El bin n de width_bucket(x, 8, 18, 20) cubre [8 + (n-1)*0.5, 8 + n*0.5) en escala log
ancho = (18 - 8) / 20
hist["centro_log"] = 8 + (hist["bin"] - 0.5) * ancho
hist["centro_pesos"] = np.expm1(hist["centro_log"])

fig, ax = plt.subplots(figsize=(10, 5))
colores = {"Hogar no pobre": "#2980b9", "Hogar pobre (no indigente)": "#d35400", "Hogar en indigencia": "#c0392b"}
for grupo, sub in hist.groupby("condicion_hogar"):
    sub = sub[(sub["bin"] >= 1) & (sub["bin"] <= 20)].sort_values("bin")
    # se normaliza cada grupo a su propio total: interesa la forma, no el tamaño del grupo
    ax.plot(sub["centro_log"], sub["personas"] / sub["personas"].sum() * 100,
            marker="o", label=grupo, color=colores.get(grupo))

ticks = np.arange(8, 18.1, 1)
ax.set_xticks(ticks)
ax.set_xticklabels([f"{np.expm1(t)/1_000_000:,.2f}M" for t in ticks], rotation=45)
ax.set_xlabel("Ingreso total de la persona (escala logarítmica, etiquetas en pesos)")
ax.set_ylabel("% de las personas del grupo (ponderado por fex_c)")
ax.set_title("Pregunta 8 — Distribución del ingreso personal según la condición de pobreza del hogar")
ax.legend()
plt.tight_layout()
plt.show()
````
